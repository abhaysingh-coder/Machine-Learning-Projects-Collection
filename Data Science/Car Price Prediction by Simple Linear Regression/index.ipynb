### Import Essential Libraries

```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

### Import DataSet

```python
data = pd.read_csv('Data/data.csv')
```

### Information About DataSet

```python
data.head()
```

### Checking and Processing NULL Values

```python
data.isna().sum()
```

### Separate DataSet into Independent and Dependent Variable 

```python
X = data.iloc[:,:6].values
y = data.iloc[:,6].values
```

### Encoding the DataSet

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:,2] = le.fit_transform(X[:,2])
```

### Split DataSet into Training and Testing DataSet

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
X_train = ss.fit_transform(X_train)
X_test = ss.transform(X_test)
```

### Apply Linerar Regression Algorithm

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
```

### Testing the Model

```python
y_pred = model.predict(X_test)
y_pred
```

### Predict According to user Requirement

```python
Car_Age = 21
KM_Driven = 10000.0
Fuel_Type = 1
Owner_Type = 1
Mileage = 15.0
Engine_Size = 1500
Price = model.predict(ss.fit_transform([[Car_Age, KM_Driven, Fuel_Type, Owner_Type, Mileage, Engine_Size]]))
print('Predicted Price:', Price)
```

```python
print(f'The Predict Price for Car of\nCar_Age       {Car_Age},\nKM_Driven     {KM_Driven},\nFuel_Type     {Fuel_Type},\nOwner_Type    {Owner_Type},\nMileage       {Mileage},\nEngine_Size   {Engine_Size},\nPrice         {Price},\n')
```

### Save the Model

```python
import joblib
joblib.dump(model, '../../Model/Car Price Prediction by Simple Linear Regression.pkl')
print('Model Saved Succesfully')
```
